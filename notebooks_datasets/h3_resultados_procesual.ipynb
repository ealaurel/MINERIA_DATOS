### Fórmula de Calificación

$$Calificación = \left( \frac{R_{correctas}}{26} \times 5 \times 0.75 \right) + \left( \frac{1}{Ranking} \times 5 \times 0.25 \right)$$

**Donde:**
* **$R_{correctas}$**: Número de respuestas correctas.
* **$Ranking$**: Ranking obtenido en Blooket (Oro).
* **26**: Valor de referencia basado en el percentil 90.
* **5**: Puntaje máximo de la evaluación.
* **0.75 / 0.25**: Pesos porcentuales (75% conocimientos, 25% ranking).

> *Aclaración:* Se tomó el **percentil 90 (26)** en lugar del máximo total (44) para asegurar una ponderación más justa y evitar que la calificación global se viera afectada negativamente.

Nota sobre el Valor de Referencia (26)
Es importante aclarar que para el cálculo de las respuestas correctas se utilizó el valor 26 como denominador. Este número corresponde aproximadamente al percentil 90 del total de respuestas por estudiante.

¿Por qué no usar el máximo real?
Aunque el total de respuestas máximo registrado fue de 44, utilizar ese número como base habría bajado injustamente la calificación de la mayoría. Al usar el percentil 90 (26), logramos una mejor ponderación, permitiendo que los estudiantes con un alto desempeño alcancen la nota máxima sin ser penalizados por valores atípicos excesivamente altos.
